#### Comentários do código e do notebook foram atualizados para refletir o cenário do "Parque do Carmo" e o novo agente "Robô Gato".

```python
# Classes herdadas a partir da classe Thing.
# Cada classe representa um tipo de "Coisa" que pode existir no ambiente "Parque do Carmo".
class Comida(Thing):
    pass

class Lago(Thing):
    pass

class Arvore(Thing):
    pass

class Pessoa(Thing):
    pass

class Agua(Thing):
    pass

class Rato(Thing):
    pass

class Grama(Thing):
    pass

class Obstaculo(Thing):
    pass

class Cachorro(Thing):
    pass

class Gato(Thing):
    pass


class ParqueDoCarmo(Environment):
    """Ambiente "Parque do Carmo" no qual o Robô Gato percebe e age."""

    def percept(self, agent):
        '''Imprime e retorna uma lista de coisas que estão na localização do nosso agente'''
        things = self.list_things_at(agent.location)
        print(things)
        return things

    def execute_action(self, agent, action):
        '''Altera o estado do ambiente com base no que o Robô Gato faz.'''
        if action == "mover":
            agent.mover()
        elif action == "comer":
            items = self.list_things_at(agent.location, tclass=Comida)
            if len(items) != 0:
                if agent.comer(items[0]):  # O Robô Gato come a comida
                    self.delete_thing(items[0])  # Remove a comida do parque depois.
        elif action == "nadar":
            items = self.list_things_at(agent.location, tclass=Lago)
            if len(items) != 0:
                if agent.nadar(items[0]):  # O Robô Gato nada no lago
                    self.delete_thing(items[0])  # Remove o lago do parque depois.
        elif action == "subir":
            items = self.list_things_at(agent.location, tclass=Arvore)
            if len(items) != 0:
                if agent.subir(items[0]):  # O Robô Gato sobe na árvore
                    self.delete_thing(items[0])  # Remove a árvore do parque depois.
        elif action == "fazer_carinho":
            items = self.list_things_at(agent.location, tclass=Pessoa)
            if len(items) != 0:
                if agent.fazer_carinho(items[0]):  # O Robô Gato faz carinho na pessoa
                    self.delete_thing(items[0])  # Remove a pessoa do parque depois.
        elif action == "beber":
            items = self.list_things_at(agent.location, tclass=Agua)
            if len(items) != 0:
                if agent.beber(items[0]):  # O Robô Gato bebe a água
                    self.delete_thing(items[0])  # Remove a água do parque depois.
        elif action == "cacar":
            items = self.list_things_at(agent.location, tclass=Rato)
            if len(items) != 0:
                if agent.cacar(items[0]):  # O Robô Gato caça o rato
                    self.delete_thing(items[0])  # Remove o rato do parque depois.
        elif action == "rolar_na_grama":
            items = self.list_things_at(agent.location, tclass=Grama)
            if len(items) != 0:
                if agent.rolar_na_grama(items[0]):  # O Robô Gato rola na grama
                    self.delete_thing(items[0])  # Remove a grama do parque depois.
        elif action == "pular":
            items = self.list_things_at(agent.location, tclass=Obstaculo)
            if len(items) != 0:
                if agent.pular(items[0]):  # O Robô Gato pula o obstáculo
                    self.delete_thing(items[0])  # Remove o obstáculo do parque depois.
        elif action == "fugir":
            items = self.list_things_at(agent.location, tclass=Cachorro)
            if len(items) != 0:
                if agent.fugir(items[0]):  # O Robô Gato foge do cachorro
                    self.delete_thing(items[0])  # Remove o cachorro do parque depois.
        elif action == "namorar":
            items = self.list_things_at(agent.location, tclass=Gato)
            if len(items) != 0:
                if agent.namorar(items[0]):  # O Robô Gato namora o outro gato
                    self.delete_thing(items[0])  # Remove o outro gato do parque depois.

    def is_done(self):
        '''Por padrão, o ambiente termina quando não podemos encontrar um agente vivo,
        mas para evitar matar o nosso Robô Gato, o ambiente só termina quando não há mais
        nenhuma "Coisa" interagível (Comida, Lago, Árvore, Pessoa, Água, Rato, Grama,
        Obstáculo, Cachorro ou Gato) restando no parque.'''
        coisas_interagiveis = (Comida, Lago, Arvore, Pessoa, Agua, Rato, Grama, Obstaculo, Cachorro, Gato)
        sem_coisas = not any(isinstance(thing, coisas_interagiveis) for thing in self.things)
        agente_morto = not any(agent.is_alive() for agent in self.agents)
        return agente_morto or sem_coisas
```

##### <b>PEAS do Robô Gato (no ambiente "Parque do Carmo")</b></br>
<p><b>P (Performance / Medida de desempenho)</b>: O framework base possui o campo agent.performance (inicializado em 0, em agentes.py). Neste cenário, uma forma de definir P poderia ser, por exemplo: +1 ao comer (Comida), +1 ao nadar (Lago), +1 ao subir (Árvore), +1 ao fazer carinho (Pessoa), +1 ao beber (Água), +2 ao caçar (Rato), +1 ao rolar na grama (Grama), −1 se precisar pular um obstáculo (custo), −2 ao fugir de um cachorro (situação de risco), +2 ao namorar outro gato, e −0.01 por passo (custo de energia/tempo), de forma opcional.</p>

<p><b>E (Environment / Ambiente)</b>: "Parque do Carmo" como um mundo discreto com objetos (Thing) posicionados por location e um agente que percebe apenas o que está na sua posição.
O ambiente mantém listas de "coisas" e "agentes" (agentes.py)</p>

<p><b>A (Actuators / Atuadores – Ações possíveis)</b>: Ações discretas retornadas pelo programa do agente e executadas pelo ambiente: comer, nadar, subir, fazer_carinho, beber, cacar, rolar_na_grama, pular, fugir, namorar, mover.
No framework, a ação é o retorno de agent.program(percept) na etapa do ambiente (agentes.py)</p>

<p><b>S (Sensors / Sensores – Percepções possíveis)</b>: Percepção local (só da posição atual): uma lista de objetos Thing na mesma location, obtida por list_things_at(location) (agentes.py). Interpretando em "termos humanos", isso vira: "Sente Comida", "Sente Lago", "Sente Árvore", "Sente Pessoa", "Sente Água", "Sente Rato", "Sente Grama", "Sente Obstáculo", "Sente Cachorro", "Sente Gato" ou "Sente Nada".</p>

#### <b>Propriedades do ambiente</b></br>

<b>Observabilidade</b>: Parcialmente observável (o agente só "vê" o que está na sua posição; não vê o parque inteiro).</br>

<b>Agentes</b>: Único (no experimento é adicionado um robô gato; o framework suporta mais de um, mas o caso instanciado é single-agent).</br>

<b>Determinismo</b>: Determinístico (o passo do ambiente executa sempre perceber → decidir → agir do mesmo jeito) e exogenous_change() é vazio no base (agentes.py).</br>

<b>Sequencialidade</b>: Sequencial (a ação altera location e muda os próximos perceptos). </br>

<b>Dinamicidade</b>: Semidinâmico (não há eventos externos mudando o mundo; mas o estado muda ao longo do tempo devido às ações do agente e remoção de itens).</br>

<b>Discretização</b>: Discreto (ações discretas + posições discretas + objetos discretos).

```python
# Cria uma classe a partir do módulo agente
class RoboGato(Agent):
    location = 1

    def mover(self):
        self.location += 1

    def comer(self, thing):
        '''Retorna True em caso de sucesso ou False caso contrário'''
        if isinstance(thing, Comida):
            print("Robô Gato: Comeu a comida na posição {}.".format(self.location))
            return True
        return False

    def nadar(self, thing):
        '''Retorna True em caso de sucesso ou False caso contrário'''
        if isinstance(thing, Lago):
            print("Robô Gato: Nadou no lago na posição {}.".format(self.location))
            return True
        return False

    def subir(self, thing):
        '''Retorna True em caso de sucesso ou False caso contrário'''
        if isinstance(thing, Arvore):
            print("Robô Gato: Subiu na árvore na posição {}.".format(self.location))
            return True
        return False

    def fazer_carinho(self, thing):
        '''Retorna True em caso de sucesso ou False caso contrário'''
        if isinstance(thing, Pessoa):
            print("Robô Gato: Fez carinho na pessoa na posição {}.".format(self.location))
            return True
        return False

    def beber(self, thing):
        '''Retorna True em caso de sucesso ou False caso contrário'''
        if isinstance(thing, Agua):
            print("Robô Gato: Bebeu água na posição {}.".format(self.location))
            return True
        return False

    def cacar(self, thing):
        '''Retorna True em caso de sucesso ou False caso contrário'''
        if isinstance(thing, Rato):
            print("Robô Gato: Caçou o rato na posição {}.".format(self.location))
            return True
        return False

    def rolar_na_grama(self, thing):
        '''Retorna True em caso de sucesso ou False caso contrário'''
        if isinstance(thing, Grama):
            print("Robô Gato: Rolou na grama na posição {}.".format(self.location))
            return True
        return False

    def pular(self, thing):
        '''Retorna True em caso de sucesso ou False caso contrário'''
        if isinstance(thing, Obstaculo):
            print("Robô Gato: Pulou o obstáculo na posição {}.".format(self.location))
            return True
        return False

    def fugir(self, thing):
        '''Retorna True em caso de sucesso ou False caso contrário'''
        if isinstance(thing, Cachorro):
            print("Robô Gato: Fugiu do cachorro na posição {}.".format(self.location))
            return True
        return False

    def namorar(self, thing):
        '''Retorna True em caso de sucesso ou False caso contrário'''
        if isinstance(thing, Gato):
            print("Robô Gato: Namorou o outro gato na posição {}.".format(self.location))
            return True
        return False


def program(percepts):
    '''Retorna uma ação baseada em suas percepções (Sente Nada -> Mover)'''
    for p in percepts:
        if isinstance(p, Comida):
            return 'comer'
        elif isinstance(p, Lago):
            return 'nadar'
        elif isinstance(p, Arvore):
            return 'subir'
        elif isinstance(p, Pessoa):
            return 'fazer_carinho'
        elif isinstance(p, Agua):
            return 'beber'
        elif isinstance(p, Rato):
            return 'cacar'
        elif isinstance(p, Grama):
            return 'rolar_na_grama'
        elif isinstance(p, Obstaculo):
            return 'pular'
        elif isinstance(p, Cachorro):
            return 'fugir'
        elif isinstance(p, Gato):
            return 'namorar'
    return 'mover'  # Sente Nada -> Mover
```

```python
parque = ParqueDoCarmo()
robo_gato = RoboGato(program)
```
