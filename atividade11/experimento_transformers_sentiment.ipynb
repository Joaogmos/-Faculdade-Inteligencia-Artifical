````python
# Importando as bibliotecas necessárias
import re
import pandas as pd
import numpy as np
import torch
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
import matplotlib.pyplot as plt
from transformers import BertTokenizer, BertForSequenceClassification, Trainer, TrainingArguments, set_seed
from datasets import Dataset, DatasetDict

# Semente fixa para que a divisão dos dados, a inicialização da camada de classificação
# e o embaralhamento dos lotes sejam reprodutíveis entre execuções
set_seed(42)
````

````python
# Função para pré-processamento do texto
# Dicionário de gírias/abreviações comuns em redes sociais -> forma por extenso.
# A troca é feita por palavra inteira (\b ... \b). No código original, text.replace("q", "que")
# trocava TODA letra "q" do texto (ex.: "quando" virava "queuando"), por isso foi corrigido.
GIRIAS = {
    "vc": "você", "vcs": "vocês", "tb": "também", "tbm": "também", "q": "que",
    "pq": "porque", "mto": "muito", "mt": "muito", "dnv": "de novo", "n": "não",
    "fds": "fim de semana", "blz": "beleza", "td": "tudo", "msm": "mesmo", "hj": "hoje",
}

# Faixas Unicode dos emojis e símbolos gráficos mais usados
EMOJIS = re.compile("[\U0001F000-\U0001FAFF\U00002600-\U000027BF\U0000FE0F\U0000200D]+")

def preprocess_text(text):
    """
    Converte o texto para minúsculas, remove URLs, hashtags, menções e emojis
    e expande gírias/abreviações por extenso.
    """
    # Convertendo para minúsculas
    text = text.lower()

    # Remover links, hashtags e menções. O original usava str.replace com padrões de regex,
    # mas str.replace procura o texto literal "http\S+" e nunca remove nada; aqui usamos re.sub.
    text = re.sub(r"http\S+", "", text)   # Remove URLs
    text = re.sub(r"#\S+", "", text)      # Remove hashtags
    text = re.sub(r"@\S+", "", text)      # Remove menções (@usuario)
    text = EMOJIS.sub(" ", text)           # Remove emojis

    # Expandindo gírias e abreviações (só quando aparecem como palavra isolada)
    for giria, extenso in GIRIAS.items():
        text = re.sub(rf"\b{giria}\b", extenso, text)

    # Normalizando espaços repetidos
    return re.sub(r"\s+", " ", text).strip()

# Teste rápido da função
print(preprocess_text("Vc viu q a entrega atrasou dnv? Quando chega?? 😡 #raiva @loja http://x.com"))
````

````python
# Dividindo o dataset em treino (70%), validação (15%) e teste (15%)
train_texts, temp_texts, train_labels, temp_labels = train_test_split(df['Mensagem'], df['Sentimento'], test_size=0.3, random_state=42)
val_texts, test_texts, val_labels, test_labels = train_test_split(temp_texts, temp_labels, test_size=0.5, random_state=42)
print(f"Treino: {len(train_texts)} | Validação: {len(val_texts)} | Teste: {len(test_texts)}")
````

````python
# Os rótulos são "Positivo", "Negativo" ou "Neutro"; o modelo trabalha com inteiros
label_mapping = {"Positivo": 1, "Negativo": 0, "Neutro": 2}
# Mapeamento inverso (inteiro -> nome da classe), usado para interpretar as previsões.
# O notebook original usava a lista ['Positivo', 'Negativo', 'Neutro'] para isso, o que trocava
# Positivo com Negativo (o índice 0 é Negativo e o 1 é Positivo). Ver predict_sentiment mais abaixo.
id2label = {v: k for k, v in label_mapping.items()}
print(id2label)

train_labels = [label_mapping[label] for label in train_labels]
val_labels = [label_mapping[label] for label in val_labels]
test_labels = [label_mapping[label] for label in test_labels]
````

````python
# Carregando o tokenizer do BERTimbau
tokenizer = BertTokenizer.from_pretrained('neuralmind/bert-base-portuguese-cased')
````

````python
# Tokenizando as mensagens
train_encodings = tokenizer(list(train_texts), truncation=True, padding=True, max_length=128, return_tensors='pt')
val_encodings = tokenizer(list(val_texts), truncation=True, padding=True, max_length=128, return_tensors='pt')
test_encodings = tokenizer(list(test_texts), truncation=True, padding=True, max_length=128, return_tensors='pt')

# Convertendo para Dataset
train_dataset = Dataset.from_dict({ 'input_ids': train_encodings['input_ids'], 'attention_mask': train_encodings['attention_mask'], 'labels': train_labels })
val_dataset = Dataset.from_dict({ 'input_ids': val_encodings['input_ids'], 'attention_mask': val_encodings['attention_mask'], 'labels': val_labels })
test_dataset = Dataset.from_dict({ 'input_ids': test_encodings['input_ids'], 'attention_mask': test_encodings['attention_mask'], 'labels': test_labels })
````

````python
# Criando o modelo BERTimbau para classificação (3 classes).
# id2label/label2id ficam gravados no config.json do modelo salvo, então o modelo carregado
# depois já sabe o nome de cada classe.
model = BertForSequenceClassification.from_pretrained(
    'neuralmind/bert-base-portuguese-cased',
    num_labels=3,
    id2label=id2label,
    label2id=label_mapping,
)
````

#### Questão 2 do relatório: número de épocas > 15

Foi usado **`num_train_epochs=20`**. Além disso:
- `eval_strategy="epoch"` e `logging_strategy="epoch"` registram o loss de treino e as métricas de validação ao fim de cada época, que é o que os gráficos pedem (no original nada era registrado por época e a função `plot_metrics` não tinha dados para plotar).
- `warmup_steps=500` foi trocado por `warmup_steps=0.1` (no Transformers 5.x um valor entre 0 e 1 é interpretado como fração do total de passos). Com ~108 mensagens de treino e lote 16, cada época tem 7 passos, ou 140 passos em 20 épocas. Com 500 passos de aquecimento a taxa de aprendizado nunca chegaria ao valor configurado (pararia em 28% dele). Com 0.1, os primeiros 14 passos fazem o aquecimento.
- `save_strategy="no"` evita gravar um checkpoint de 420 MB por época; o modelo final é salvo manualmente mais abaixo.

````python
# Definindo os parâmetros de treinamento
training_args = TrainingArguments(
    output_dir='./results',          # Pasta para armazenar os resultados
    num_train_epochs=20,             # Número de épocas de treinamento (> 15, pedido na atividade)
    per_device_train_batch_size=16,  # Tamanho do lote de treinamento
    per_device_eval_batch_size=64,   # Tamanho do lote de avaliação
    warmup_steps=0.1,                # Valor < 1 = fração dos passos: 10% do treino para aquecimento da taxa de aprendizado
    weight_decay=0.01,               # Decaimento de peso
    eval_strategy='epoch',           # Avalia ao fim de cada época
    logging_strategy='epoch',        # Registra o loss de treino ao fim de cada época
    save_strategy='no',              # Não grava checkpoints intermediários
    report_to='none',                # Sem integração com wandb/tensorboard
    seed=42,
)
print("num_train_epochs =", training_args.num_train_epochs)
````

````python
from sklearn.metrics import accuracy_score
import numpy as np

# Definindo a função de métrica
def compute_metrics(p):
    predictions, labels = p
    predictions = np.argmax(predictions, axis=1)  # Convertendo logits para classes
    return {"accuracy": accuracy_score(labels, predictions)}
````

````python
# O DataCollatorWithPadding é utilizado para adicionar automaticamente o token de padding ([PAD]) 
# nas sequências de texto de um batch, garantindo que todas as sequências tenham o mesmo comprimento. 
# Ele utiliza o tokenizer fornecido para calcular o padding necessário com base nas configurações do modelo. 
# Esse processo é essencial porque os modelos de NLP, como o BERT, exigem entradas com o mesmo tamanho. 
# O collator facilita o preparo dos dados durante o treinamento, sem a necessidade de pré-processamento manual 
# do padding. Ele é especialmente útil em batches com sequências de comprimento variável.
from transformers import DataCollatorWithPadding

data_collator = DataCollatorWithPadding(tokenizer=tokenizer)
````

````python
# Definindo a função Trainer.
# eval_dataset recebe um dicionário com treino e validação: assim, ao fim de cada época, o Trainer
# calcula loss e acurácia nos dois conjuntos (chaves eval_treino_* e eval_validacao_* no histórico),
# o que permite plotar a curva de acurácia de treinamento além da de validação.
trainer = Trainer(
    model=model,                         # O modelo
    args=training_args,                  # Argumentos de treinamento
    train_dataset=train_dataset,         # Dataset de treinamento
    eval_dataset={'treino': train_dataset, 'validacao': val_dataset},  # Avaliados a cada época
    data_collator=data_collator,         # Usando DataCollatorWithPadding
    compute_metrics=compute_metrics      # Passando a função de métrica
)
````

````python
# Treinando o modelo (a tabela mostra, por época, o loss de treino e as métricas de treino/validação)
trainer.train()
````

#### Questão 3 do relatório: loss de treino em cada época

A tabela abaixo é montada a partir de `trainer.state.log_history`, o registro que o `Trainer` guarda durante o treino.

````python
# Organizando o histórico de treino por época
hist = pd.DataFrame(trainer.state.log_history)
por_epoca = (
    hist.groupby('epoch')
        .agg({'loss': 'first', 'eval_treino_loss': 'first', 'eval_treino_accuracy': 'first',
              'eval_validacao_loss': 'first', 'eval_validacao_accuracy': 'first'})
        .dropna(subset=['loss'])
        .reset_index()
)
por_epoca.columns = ['Época', 'Loss de treino', 'Loss (treino, modo avaliação)', 'Acurácia treino',
                     'Loss validação', 'Acurácia validação']
por_epoca['Época'] = por_epoca['Época'].round().astype(int)
pd.set_option('display.precision', 4)
display(por_epoca)
por_epoca.to_csv('historico_epocas.csv', index=False, encoding='utf-8')
````

#### Questão 4 do relatório: resultados de avaliação em treino, validação e teste

````python
# Avaliando o modelo no treino, na validação e no teste.
# metric_key_prefix='eval' mantém as chaves no formato eval_loss / eval_accuracy usado abaixo.
train_results = trainer.evaluate(train_dataset, metric_key_prefix='eval')
val_results = trainer.evaluate(val_dataset, metric_key_prefix='eval')
test_results = trainer.evaluate(test_dataset, metric_key_prefix='eval')
````

````python
# Verificando o conteúdo dos resultados de avaliação
print(train_results)
print(val_results)
print(test_results)
````

````python
# Exibindo os resultados de avaliação
print(f"Treinamento - Acurácia: {train_results['eval_accuracy']:.4f} | Loss: {train_results['eval_loss']:.4f}")
print(f"Validação   - Acurácia: {val_results['eval_accuracy']:.4f} | Loss: {val_results['eval_loss']:.4f}")
print(f"Teste       - Acurácia: {test_results['eval_accuracy']:.4f} | Loss: {test_results['eval_loss']:.4f}")
````

````python
# Gerando os gráficos de Loss e Acurácia
def plot_metrics(metrics):
    epocas = metrics["epoch"]

    # Plotando o loss de treino e validação
    plt.figure(figsize=(12, 6))
    plt.plot(epocas, metrics["train_loss"], marker="o", label="Loss de Treinamento")
    plt.plot(epocas, metrics["eval_loss"], marker="o", label="Loss de Validação")
    plt.xlabel("Épocas")
    plt.ylabel("Loss")
    plt.xticks(epocas)
    plt.legend()
    plt.grid(alpha=0.3)
    plt.title("Loss de Treinamento e Validação")
    plt.savefig("grafico_loss.png", dpi=120, bbox_inches="tight")
    plt.show()

    # Plotando a acurácia de treino e validação
    plt.figure(figsize=(12, 6))
    plt.plot(epocas, metrics["train_accuracy"], marker="o", label="Acurácia de Treinamento")
    plt.plot(epocas, metrics["eval_accuracy"], marker="o", label="Acurácia de Validação")
    plt.xlabel("Épocas")
    plt.ylabel("Acurácia")
    plt.xticks(epocas)
    plt.ylim(0, 1.05)
    plt.legend()
    plt.grid(alpha=0.3)
    plt.title("Acurácia de Treinamento e Validação")
    plt.savefig("grafico_acuracia.png", dpi=120, bbox_inches="tight")
    plt.show()
````

````python
# Questão 5 do relatório: montando o dicionário de métricas por época e chamando plot_metrics
# (no notebook original a função era definida mas nunca chamada)
metrics = {
    "epoch": por_epoca['Época'].tolist(),
    "train_loss": por_epoca['Loss de treino'].tolist(),
    "eval_loss": por_epoca['Loss validação'].tolist(),
    "train_accuracy": por_epoca['Acurácia treino'].tolist(),
    "eval_accuracy": por_epoca['Acurácia validação'].tolist(),
}
plot_metrics(metrics)
````

````python
# Gerando previsões para o conjunto de teste
test_results = trainer.predict(test_dataset)

# Obtendo as previsões (probabilidades ou logits)
y_pred = test_results.predictions
y_pred = np.argmax(y_pred, axis=1)  # Convertendo logits para classe
````

````python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

# Questão 6 do relatório: matriz de confusão no conjunto de teste
test_results = trainer.predict(test_dataset)
y_pred = np.argmax(test_results.predictions, axis=1)

# Calculando a matriz de confusão. labels=[1, 0, 2] fixa a ordem Positivo, Negativo, Neutro,
# para que os rótulos dos eixos correspondam às linhas/colunas. No original a matriz saía na ordem
# 0, 1, 2 (Negativo, Positivo, Neutro) mas os eixos diziam Positivo, Negativo, Neutro.
ordem = [1, 0, 2]
nomes = [id2label[i] for i in ordem]
cm = confusion_matrix(test_labels, y_pred, labels=ordem)

# Plotando a matriz de confusão
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=nomes, yticklabels=nomes)
plt.xlabel("Predição")
plt.ylabel("Real")
plt.title("Matriz de Confusão - Teste")
plt.savefig("matriz_confusao_teste.png", dpi=120, bbox_inches="tight")
plt.show()

print(classification_report(test_labels, y_pred, labels=ordem, target_names=nomes, zero_division=0))
````

````python
# Movendo o modelo carregado para o dispositivo e colocando em modo de avaliação (desliga o dropout)
modelo_carregado.to(device)
modelo_carregado.eval()
````

````python
# Função para testar o modelo com uma nova mensagem
def predict_sentiment(message, modelo=None, tok=None):
    modelo = modelo or modelo_carregado
    tok = tok or tokenizer_carregado
    message = preprocess_text(message)  # Pré-processando a mensagem
    inputs = tok(message, return_tensors='pt', truncation=True, padding=True, max_length=128)

    # Movendo os dados para o mesmo dispositivo do modelo
    inputs = {key: value.to(device) for key, value in inputs.items()}

    # Realizando a predição
    with torch.no_grad():
        logits = modelo(**inputs).logits

    # Determinando a classe predita. Usa o id2label gravado no modelo (0=Negativo, 1=Positivo,
    # 2=Neutro). A lista fixa ['Positivo', 'Negativo', 'Neutro'] do original invertia Positivo e Negativo.
    prediction = torch.argmax(logits, dim=-1).item()
    return modelo.config.id2label[prediction]
````

````python
# Testando com a mensagem do notebook original
new_message = "Não gostei desta roupa!"
predicted_sentiment = predict_sentiment(new_message)
print(f'A previsão de sentimento para a mensagem é: {predicted_sentiment}')
````

````python
# A partir deste ponto, faça o uso de mais 5 mensagens e verifique o sentimento de cada uma delas
# Mensagens informais escritas pelo grupo (com gírias, abreviações e emojis), nenhuma delas usada no treino.
# Para cada uma guardamos a classe real (atribuída pelo grupo) e a classe prevista pelo modelo carregado.
resultados_testes = []

def testar(mensagem, classe_real):
    prevista = predict_sentiment(mensagem)
    resultados_testes.append({'Mensagem': mensagem, 'Texto após pré-processamento': preprocess_text(mensagem),
                              'Classe real': classe_real, 'Classe prevista': prevista,
                              'Acertou?': 'Sim' if prevista == classe_real else 'Não'})
    print(f'Mensagem: {mensagem}')
    print(f'Classe real: {classe_real} | Classe prevista: {prevista}')
````

````python
# Testando com 1a nova mensagem
testar("mano o show ontem foi top demais, curti mto 😍🔥", "Positivo")
````
